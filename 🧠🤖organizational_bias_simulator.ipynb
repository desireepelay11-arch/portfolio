# Modelando "El fin del sesgo" (Jessica Nordell📚) en Sistemas de IA Locales

Este simulador traduce los principios de diseño conductual y neurociencia del libro *"El fin del sesgo"* de Jessica Nordell a variables lógicas medibles.

### El problema de negocio:
Cuando las organizaciones configuran herramientas de IA locales (como asistentes integrados en Workday, SuccessFactors o modelos LLM privados) y las entrenan con datos históricos de la empresa, corren un riesgo crítico: **automatizar y escalar los sesgos cognitivos históricos del liderazgo a nivel de software.**

A continuación, modelamos el impacto sistémico de un liderazgo en "piloto automático" versus un sistema con "intervenciones estructuradas" a lo largo de un ciclo anual (3 trimestres).


## 1. Definición del Entorno Organizacional

En esta sección definimos la clase `EquipoTrabajo` con tres métricas clave que impactan directamente en el rendimiento del negocio:
* **Diversidad Cognitiva:** La variedad de perspectivas y enfoques de resolución de problemas en el equipo.
* **Capacidad de Innovación:** La velocidad y calidad con la que el sistema genera soluciones disruptivas.
* **Retención de Talento:** El indicador de salud cultural y seguridad psicológica.


## 2. Simulación de Escenarios: Sesgo Automático vs. Intervención Estructural

Aquí ejecutamos la simulación comparativa evaluando dos comportamientos del sistema:

1. **Escenario A (Sesgo Automático / Piloto Automático):** La IA replica el *Sesgo de Afinidad* (homofilia social en promociones) y la *Evaluación Sesgada* (penalización desproporcionada del error en perfiles no tradicionales). El algoritmo degrada la diversidad y dispara la fuga de talento.
2. **Escenario B (Enfoque J. Nordell):** Se introducen *Intervenciones Estructurales* (checklists obligatorios, pausas cognitivas en la toma de decisiones y auditorías de criterios). El sistema mitiga la subjetividad del liderazgo humano, estabilizando las métricas de innovación y retención.


```python
import random

class EquipoTrabajo:
    def __init__(self, diversidad_inicial):
        self.diversidad = diversidad_inicial  # Porcentaje de diversidad cognitiva/perfiles
        self.innovacion = 50                 # Escala de 0 a 100
        self.retencion_talento = 80          # Escala de 0 a 100

    def simular_ciclo_con_sesgo(self):
        # 1. Sesgo de Afinidad: Reduce la diversidad en las promociones
        self.diversidad -= random.randint(2, 5)
        # 2. Evaluación Sesgada: Frustra al equipo y baja la retención
        self.retencion_talento -= random.randint(3, 7)
        # Menos diversidad e incentivos penalizan la innovación
        self.innovacion -= (100 - self.diversidad) * 0.1 + random.randint(1, 4)

        # Mantener límites lógicos entre 0 y 100
        self.diversidad = max(10, self.diversidad)
        self.innovacion = max(10, self.innovacion)
        self.retencion_talento = max(10, self.retencion_talento)

    def simular_ciclo_con_intervencion(self):
        # Aplicando metodologías del libro de Nordell (Checklists, pausa consciente)
        self.diversidad += random.randint(1, 3)
        self.retencion_talento += random.randint(2, 5)
        self.innovacion += random.randint(3, 6)

        self.diversidad = min(100, self.diversidad)
        self.innovacion = min(100, self.innovacion)
        self.retencion_talento = min(100, self.retencion_talento)

```
