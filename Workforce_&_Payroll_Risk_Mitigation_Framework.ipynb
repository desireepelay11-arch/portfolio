# 📊 Workforce & Payroll Risk Mitigation Framework
## 🏭 Reporte de Impacto Financiero en OPEX e Ineficiencias de Nómina (Planta AR)

### 📌 Contexto del Negocio
En entornos industriales de consumo masivo, el ausentismo laboral no es solo un indicador administrativo; es una fuga directa de eficiencia en la cadena de suministro. Este framework traduce métricas operativas de capital humano a **lenguaje financiero ($)** para facilitar la toma de decisiones del C-Level y la alineación con los CoEs (Centros de Excelencia).

### 🔍 Hallazgos Críticos Detectados
1. **El Foco de Riesgo:** El área de **Logistics & Inventory** presenta un ausentismo real del **8.5%**, duplicando el objetivo máximo de la planta (4%).
2. **El Costo Oculto:** Esta ineficiencia genera un sobrecosto inmediato en el **OPEX de Horas Extras**, debido a la necesidad de cubrir turnos críticos al 50% de recargo de ley para evitar paradas de línea.
3. **El Pipeline de Nómina:** Identificamos que los retrasos en las novedades ocurren por procesos manuales descentralizados en las 48 horas previas al cierre de liquidación en Workday/SAP.

### 💡 Solución Estratégica & Retorno de Inversión (ROI)
Implementación de un **"War Room" presencial uno a uno** con los supervisores de turno con historial de desvíos. Al combinar soporte técnico operativo con concientización cultural (Growth Mindset), proyectamos reducir el ausentismo de Logística al **4.5%**, generando un **ahorro anualizado de USD 50,005.20** para el P&L de la compañía.


## 📈 Executive Dashboard: Diagnóstico Financiero y Mitigación de Riesgos en Planta

A continuación se desglosan los KPIs obtenidos a través del modelo analítico, consolidando la información de nómina, ausentismo e impacto en el presupuesto operativo (OPEX).

### 🚨 KPI 1: Tasa de Ausentismo vs. Objetivo Corporativo (Gráfico 1)
*   **Métrica Detectada:** El área de **Logistics & Inventory** registra una tasa de ausentismo real del **8.5%**.
*   **Diagnóstico:** Este volumen duplica el objetivo máximo permitido (**4.0%**), identificando al sector logístico como la principal zona de riesgo operativo y fatiga laboral dentro de la planta.
*   **Acción de Red:** Colaboración directa con el *Medical & Safety CoE* para auditar las causas raíz de las licencias médicas y analizar la ergonomía en los turnos rotativos.

### 💸 KPI 2: Pérdida Total e Impacto en OPEX Mensual (Gráfico 2)
*   **Métrica Detectada:** El desvío de ausentismo en Logística e Inventario se traduce en un costo de **USD 17,907.12 mensuales**.
*   **Diagnóstico Financiero:** Esta fuga de capital no es puramente administrativa; contempla la nómina hundida no productiva combinada con la inyección de **Horas Extras (pagadas al 1.50% de recargo)** necesarias para cubrir los puestos vacíos y asegurar que la cadena de suministro y distribución de consumo masivo no se detenga.

### 🔮 KPI 3: Simulación de Retorno de Inversión / ROI (Gráfico 3)
*   **Métrica Proyectada:** Al implementar el protocolo de **"War Room" presencial uno a uno** con los supervisores del sector, se modela una reducción del ausentismo del 8.5% al **4.5%** (alineado al estándar corporativo).
*   **Impacto de la Solución:** Bajo este escenario de mitigación cultural, la pérdida mensual del área disminuye drásticamente a **USD 9,480.24**, recuperando la estabilidad del pipeline de operaciones.

---

### 🎯 Conclusión Financiera y Visión de Negocio para el P&L
La estrategia de intervención del equipo de People & Culture, soporte técnico excepcional en Workday/SAP y fortalecimiento del liderazgo proyecta un **Ahorro Neto Anualizado de USD 101,122.56** directo para el P&L de la compañía.

Este framework demuestra cómo la transformación cultural de los líderes y la agilidad en la gestión de novedades de nómina eliminan ineficiencias críticas de control de fuerza laboral, transformando un cuello de botella operativo en un valor financiero directo para el negocio.


```python
# =============================================================================
# FRAMEWORK: Workforce & Payroll Risk Mitigation (Cálculo de Gráficos Ejecutivos)
# =============================================================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. Base de datos operativa de la Planta
data_planta = {
    'Area': ['Production & Line Assembly', 'Logistics & Inventory', 'Quality & Maintenance'],
    'Headcount_FTE': [150, 80, 45],
    'Salario_Promedio_USD_Mes': [1200, 1100, 1400],
    'Ausentismo_Real_Porcentaje': [0.042, 0.085, 0.038]
}
df = pd.DataFrame(data_planta)

# 2. Reglas financieras de negocio
cargas_sociales_ar = 0.26   # 26% cargas patronales
recargo_hora_extra = 1.50   # 50% recargo por ley
tasa_cobertura_he = 0.60    # 60% de ausencias críticas se cubren con Horas Extras

# 3. Modelado matemático de costos de capital humano
df['Costo_Laboral_Mensual_FTE'] = df['Salario_Promedio_USD_Mes'] * (1 + cargas_sociales_ar)
df['FTE_Ausentes_Promedio'] = df['Headcount_FTE'] * df['Ausentismo_Real_Porcentaje']
df['Costo_Directo_Ausentismo_USD'] = df['Costo_Laboral_Mensual_FTE'] * df['FTE_Ausentes_Promedio']
df['Costo_OPEX_Horas_Extras_USD'] = df['Costo_Directo_Ausentismo_USD'] * tasa_cobertura_he * recargo_hora_extra
df['Impacto_Financiero_Total_USD'] = df['Costo_Directo_Ausentismo_USD'] + df['Costo_OPEX_Horas_Extras_USD']

# --- CONFIGURACIÓN DE LOS GRÁFICOS VISUALES PARA EL C-LEVEL ---
print("📊 Generando visualizaciones ejecutivas para la reunión de Directorio...")

# Configuración de estilo limpio y profesional
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# GRAFICO 1: Comparativa de Tasas de Ausentismo vs Objetivo (4%)
axes[0].bar(df['Area'], df['Ausentismo_Real_Porcentaje'] * 100, color=['#b0bec5', '#e53935', '#b0bec5'], width=0.5)
axes[0].axhline(y=4.0, color='red', linestyle='--', linewidth=1.5, label='Objetivo C-Level (4%)')
axes[0].set_title('Tasa de Ausentismo por Área (%)', fontsize=12, fontweight='bold', pad=10)
axes[0].set_ylabel('Porcentaje (%)')
axes[0].legend()

# GRAFICO 2: Impacto Total en el OPEX Mensual por Área
axes[1].bar(df['Area'], df['Impacto_Financiero_Total_USD'], color=['#37474f', '#d32f2f', '#37474f'], width=0.5)
axes[1].set_title('Pérdida Total en OPEX Mensual (USD)', fontsize=12, fontweight='bold', pad=10)
axes[1].set_ylabel('Costo en Dólares (USD)')

# GRAFICO 3: Impacto de tu solución (Antes vs Después en Logística)
ausentismo_antes = df.loc[df['Area'] == 'Logistics & Inventory', 'Impacto_Financiero_Total_USD'].values[0]
# Recálculo simulado al 4.5% de ausentismo con el War Room
costo_directo_nuevo = df.loc[df['Area'] == 'Logistics & Inventory', 'Costo_Laboral_Mensual_FTE'].values[0] * (80 * 0.045)
opex_he_nuevo = costo_directo_nuevo * tasa_cobertura_he * recargo_hora_extra
ausentismo_despues = costo_directo_nuevo + opex_he_nuevo

escenarios = ['Antes del War Room\n(Ausentismo 8.5%)', 'Después del War Room\n(Ausentismo 4.5%)']
valores_escenarios = [ausentismo_antes, ausentismo_despues]

axes[2].bar(escenarios, valores_escenarios, color=['#e53935', '#2e7d32'], width=0.4)
axes[2].set_title('ROI de la Intervención Cultural en Logística', fontsize=12, fontweight='bold', pad=10)
axes[2].set_ylabel('Costo Mensual del Área (USD)')

# Ajustar diseño final y mostrar en pantalla
plt.tight_layout()
plt.show()

# Mensaje sintético final de cierre
ahorro_anual = (ausentismo_antes - ausentismo_despues) * 12
print(f"\n🎯 CONCLUSIÓN FINANCIERA RECOMENDADA: La intervención cultural proyecta un ahorro neto anualizado de USD {ahorro_anual:,.2f} en eficiencia de nómina.")
```
